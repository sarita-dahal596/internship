```python
#private attribute +method
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

    def __check_balance(self):  #private method
        print("Current balance:", self.__balance)

    def show(self):
        print("Name:", self.name)
        self.__check_balance()


account = BankAccount("Ram", 5000)

account.show()
```

```python
# single level inheritance
class Car: #parent class,base class

    @staticmethod
    def start():
        print("Car started")

    @staticmethod
    def stop():
        print("Car stopped")


class ToyotaCar(Car):     #derived class ,ToyotaCar inherits from Car.#child class
    def __init__(self, name):
        self.name = name
        print("Toyota car created")


car1 = ToyotaCar("Toyota")

car1.start()  #ToyotaCar doesn't have its own stop().Python looks in the parent Car and finds it.
car1.stop()
```

```python
 #multilevel inheritance
class Car:
    def start(self):
        print("Car started")

    def stop(self):
        print("Car stopped")


class ToyotaCar(Car):
    def toyota(self):
        print("Toyota car")


class Byd(ToyotaCar):
    def electric(self):
        print("BYD electric car")

car1 = Byd()

car1.start()
car1.stop()
car1.toyota()
car1.electric()
```

```python
#multilevel
class Animal:
    @staticmethod   #don't need self,becz they don't use any object data.
    def eat():
        print("Animal eats")


class Dog(Animal):
    @staticmethod   #don't need self,becz they don't use any object data.
    def bark():
        print("Dog barks")


class Puppy(Dog):
    @staticmethod   #don't need self,becz they don't use any object data.
    def play():
        print("Puppy plays")


p1 = Puppy()

p1.eat()
p1.bark()
p1.play()
```

```python
#super .>used to access method of parent class
class Car:
    def __init__(self, type):
        self.type = type

    @staticmethod   #don't need self,becz they don't use any object data.
    def start():
        print("Car started")

    @staticmethod    #don't need self,becz they don't use any object data.
    def stop():
        print("Car stopped")


class ToyotaCar(Car):
    def __init__(self, name, type):
        super().__init__(type)   #Go to the parent class.
        self.name = name
        print("Toyota car created:", self.name)
        print("Type:", self.type)  #calls the Car constructor:


car1 = ToyotaCar("Toyota", "Electric")

car1.start()
car1.stop()
```

```python
#methodoverriding  Parent has a method → Child writes the same method again → Child's version is used.
class Car:
    def start(self):
        print("Car starts")


class ToyotaCar(Car):
    def start(self):
        print("Toyota car starts")


car1 = ToyotaCar()

car1.start()
```

```python
#polymorphism
class Dog:
    def sound(self):  #both class use same method here but they behave differently
        print("Dog barks")


class Cat:
    def sound(self):  #both class use same method here but they behave differently
        print("Cat meows")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

```python
#polymorphism
class Dog:
    def sound(self):
        print("Dog barks")


class Cat:
    def sound(self):
        print("Cat meows")


def make_sound(animal):  # function make_sound ,animal=parameter
    animal.sound()  #Call the sound() method of whatever object is inside animal.


dog = Dog()  #object of dog class
cat = Cat()  #object of cat class

make_sound(dog)  #calling the function and passing dog object
make_sound(cat)  #calling the function and passing car object
```

```python
#polymorphism
class Dog:
    def sound(self):  #same method
        print("Dog barks")


class Cat:
    def sound(self):  #same method
        print("Cat meows")


class Cow:
    def sound(self):  #same method
        print("Cow moos")


animals = [Dog(), Cat(), Cow()]

for animal in animals: #automatically takes each object one by one.
    animal.sound()
```

encapsulation

Encapsulation happens by putting data + methods inside the class and controlling access to some data using access modifiers.


```python
class Student:
    def __init__(self, name, age, marks):
        self.name = name          # Public
        self._age = age           # Protected
        self.__marks = marks      # Private

    def show_data(self):  #show_data() is the method that provides controlled access to __marks.
        print("Name:", self.name)
        print("Age:", self._age)
        print("Marks:", self.__marks)


student1 = Student("Ram", 20, 80)

student1.show_data()
```

```python
# Abstraction
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def eat(self):
        pass
    @abstractmethod
    def walk(self):
        pass

    @abstractmethod
    def speak(self):
        pass


class Dog(Animal):
    def eat(self):
        print("Dog eats")

    def walk(self):
        print("Dog walks")

    def speak(self):
        print("Dog speaks")

    def run(self):
        print("Dog runs")
```
