```python
#instance method,class method,static method
class Student:

    school = "ABC School"       # Class variable

    def __init__(self, name, age):
        self.name = name        # Instance variable
        self.age = age

    # 1. Instance method
    def display(self):  #work with a particular object
        print("Name:", self.name)
        print("Age:", self.age)

    # 2. Class method
    @classmethod   #dosen't require object
    def school_name(cls): #works with class itself
        print("School:", cls.school)

    # 3. Static method
    @staticmethod    #dosen't require object
    def welcome(): #dosen't need self
        print("Welcome to ABC School")


# Create object
s1 = Student("James", 20)

# Instance method
s1.display()

# Class method
Student.school_name()

# Static method
Student.welcome()
```

```python
#instance,class,static method
class Transport:

    vehicle = "Transport Vehicle"   # Class variable

    # Constructor
    def __init__(self, name, road):
        self.name = name             # Instance variable
        self.road = road             # Instance variable

    # Instance method
    def display(self):
        print("Name:", self.name)
        print("Road:", self.road)

    # Class method
    @classmethod
    def vehicle_name(cls):
        print("Vehicle:", cls.vehicle)

    # Static method
    @staticmethod
    def info():
        print("Enjoy your ride")


# Create object
t1 = Transport("Bike", "Mahendra Highway")

# Instance method
t1.display()

# Class method
Transport.vehicle_name()

# Static method
Transport.info()
```
