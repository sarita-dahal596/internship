Dunder methods are used to define or customize how Python objects behave with built-in operations and functions.

```python
class Student:
    def __init__(self, name):
        self.name = name

    def __str__(self):   #__self__ is dunder method.When someone tries to print my object, what should I display
        return self.name


student1 = Student("Ram")

print(student1)
```

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):   #"When the + operator is used with my objects, use this method."#__add__() is a dunder method for the + operator.
        return self.value + other.value

    def __sub__(self, other):
        return self.value - other.value

    def __mul__(self, other):
        return self.value * other.value

    def __truediv__(self, other):
        return self.value / other.value


num1 = Number(20)
num2 = Number(10)

print(num1 + num2)
print(num1 - num2)
print(num1 * num2)
print(num1 / num2)
```

Exception handling is used to handle errors in a program without stopping the entire program suddenly.

```python
try:  #code that may cause an error
    number = int(input("Enter a number: "))
    print(number)

except: #handles the error
    print("Please enter a valid number")
```

```python
#division by zero
try:
    a = 10
    b = 0

    result = a / b
    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero")
```

```python
#list index error
numbers = [10, 20, 30]

try:
    print(numbers[5])

except IndexError:
    print("Index does not exist")
```

```python
#Dictionary key error
student = {
    "name": "Ram",
    "age": 20
}

try:
    print(student["marks"])

except KeyError:
    print("Key does not exist")
```

```python
#Multiple exceptions
try:
    number = int(input("Enter number: "))
    result = 10 / number
    print(result)

except ValueError:
    print("Please enter a valid number")

except ZeroDivisionError:
    print("Cannot divide by zero")
```

```python
#try except finally
try:
    number = int(input("Enter number: "))
    print(number)

except ValueError:
    print("Invalid input")

finally:
    print("Program finished")
```

```python
try:
    number = float(input("Enter number: "))
    print(number)

except ValueError:
    print("Invalid input")

finally:
    print("Program finished")
```

```python
#rise :create/trigger the error yourself
age = 15

try:
    if age < 18:
        raise ValueError("Age must be 18 or above")

    print("You can register")

except ValueError as e:
    print(e)
```

```python
#raise with user input
try:
    age = int(input("Enter your age: "))

    if age < 0:
        raise ValueError("Age cannot be negative")

    print("Your age is:", age)

except ValueError as e:
    print("Error:", e)
```

```python
balance = 500
withdraw = 1000

try:
    if withdraw > balance:
        raise ValueError("Insufficient balance")

    balance = balance - withdraw
    print("Remaining balance:", balance)

except ValueError as e:  #This catches the ValueError created by:raise ValueError("Insufficient balance")
    print(e)  #e->error msg
```
