```python
import numpy as np
import pandas as pd

data = np.array([
    [100, -20, 30, 5000],
    [45, 90, 0, -100],
    [100, 90, 30, 2500],
    [-50, 80, 1000, 10],
    [200, -80, 0, 5000]
])

df = pd.DataFrame(data)


print(df)
```

```python
df = pd.DataFrame(
    data,
    columns=["Sales", "Discount", "Items", "Revenue"]
)

print(df)
```

```python
print(np.mean(df["Sales"]))
```

```python
print(np.max(df["Revenue"]))
```
