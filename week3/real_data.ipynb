```python
def fill_zipcode(row, df):
    if pd.isna(row["zipcode"]):
        state_data = df[df["state"] == row["state"]]
        return state_data["zipcode"].mode().iloc[0]
    return row["zipcode"]

df["zipcode"] = df.apply(fill_zipcode, axis=1, df=df)
```
