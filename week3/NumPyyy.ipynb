Mathematical functions

```python
import numpy as np

# Messy data
data = np.array([10, -50, -25.5, 900, -20, 75.8, 0, 25.5])

print("Original data:")
print(data)

# Sum
print("Sum:", np.sum(data))

# Mean
print("Mean:", np.mean(data))

# Maximum
print("Maximum:", np.max(data))

# Minimum
print("Minimum:", np.min(data))

# Median
print("Median:", np.median(data))

# Standard deviation
print("Standard deviation:", np.std(data))

# Absolute value
print("Absolute values:", np.abs(data))

# Round values
print("Rounded values:", np.round(data))
```

```python
marks = np.array([85, 90, -10, 75.5, 90, 100, 0, 65.8, -5])
print("orginal data",marks)
print("Total:", np.sum(marks))
print("Average:", np.mean(marks))
print("maximum:",np.max(marks))
print("mean:",np.mean(marks))
```

```python

expenses = np.array([
    500,
    1200,
    -300,
    750.50,
    1200,
    0,
    450.75,
    -100,
    900
])
print("original data",expenses)
print("total:",np.sum(expenses))
print("average:",np.mean(expenses))
print("lowest:",np.min(expenses))
print("highest:",np.max(expenses))
print(type(expenses))
```

```python
#above 0
clean_expenses = expenses[expenses >= 0]

print("Clean expenses:", clean_expenses)
print("Total:", np.sum(clean_expenses))
print("Average:", np.mean(clean_expenses))
```

```python
#value above a certain amount
expenses = np.array([500, 1200, 300, 750, 2000, 150, 900])

large_expenses = expenses[expenses > 1000]

print("Expenses above 1000:", large_expenses)
```

```python
#value bellow certain amount
expenses_1 = np.array([100,50,20,30,40,10,5,3])
small_expenses = expenses_1[expenses_1 <= 20]
print("Expenses below 20:", small_expenses)
```

```python
#decimal data
prices = np.array([100.567, 250.234, 75.899, 500.456])

print("Original:", prices)

rounded = np.round(prices, 1)  #.pachadiko 1 ta data dinxa

print("Rounded:", rounded)
```

```python
#absolute value
temperature = np.array([-10, 20, -5, 30, -15])

print("Original:", temperature)

result = np.abs(temperature)

print("Absolute values:", result)
```

```python
data = np.array([4, 9, 16, 25, 100])

result = np.sqrt(data)

print("Original:", data)
print("Square root:", result)
```

```python
data = np.array([2,3,2,4,5,6,5,7,8,9,7,4,10])
unique_elements =np.unique(data)
print("original data:",data)
print("unique elements",unique_elements)
```

axis operation

```python
import numpy as np

data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
np.sum(data, axis=0)
```

```python
np.sum(data, axis=1) #down to the column ko sum
```

```python
np.mean(data, axis=0) ##down to the column ko mean
```

```python
data = np.array([
    [90, 70, 50],
    [80, 50, 100]
])
np.median(data, axis=0)
```

```python
#sum using axis
import numpy as np

data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Column sum:", np.sum(data, axis=0))
print("Row sum:", np.sum(data, axis=1))
```

```python
print("Column median:", np.median(data, axis=0))
print("Row median:", np.median(data, axis=1))
```

```python
data = np.array([
    [100, 20, 30],
    [40, 90, 60],
    [50, 80, 90]
])
np.argmax(data, axis=0)
```

```python
np.argmin(data, axis=0)
```

```python
np.prod(data, axis=0)  #Multiplie column
```

```python
np.prod(data, axis=1)  #Multiplie row
```

```python
data = np.array([
    [100, 20, 30],
    [40, 90, 60],
    [50, 80, 90]
])
print(np.ptp(data))  #max value - min value here max-100,min 20
```
