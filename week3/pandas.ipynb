```python
import pandas as pd

data = {
    "Name": ["Ram", "Sita", "Hari", "Gita"],
    "Age": [20, None, 22, None],
    "City": ["Kathmandu", "Pokhara", None, "Lalitpur"]
}

df = pd.DataFrame(data)

print("Before:")
print(df)

# Check missing values
print("\nMissing values:")
print(df.isnull().sum())

# Handle missing values
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["City"] = df["City"].fillna("Unknown")

print("\nAfter:")
print(df)
```

```python
import pandas as pd

data = {
    "Category": ["Food", "Food", "Travel", "Travel", "Food"],
    "Amount": [500, 300, 1000, 500, 200]
}

df = pd.DataFrame(data)

print(df)
```

aggregation,


```python
#group by category and sum
result = df.groupby("Category")["Amount"].sum()

print(result)
```

```python
#count
print(df["Amount"].value_counts())
```

```python
#count by number
df[df['Amount']==500].value_counts()
```

```python
#total
total = df["Amount"].sum()

print("Total:", total)
```

```python
#average
average = df["Amount"].mean()

print("Average:", average)
```

```python
#highest
highest = df["Amount"].max()

print("Highest:", highest)
```

```python
#lowest
lowest = df["Amount"].min()

print("Lowest:", lowest)
```

```python
#count amount
count = df["Amount"].count()

print("Count:", count)
```

```python
#All aggregations together
result = df["Amount"].agg(["sum", "mean", "max", "min", "count"])

print(result)
```

grouping


```python
#aggegation based on group by
df.groupby('Category')['Amount'].mean()
```

```python
#group and calculate total
result = df.groupby("Category")["Amount"].sum()

print(result)
```

```python
#group and calculate category
result = df.groupby("Category").size()

print(result)
```

```python
##group and calculate amount
result = df.groupby("Amount").size()

print(result)
```

transformation

```python
#dd 100 to every Amount
df["Amount"] = df["Amount"].transform(lambda x: x + 100)

print(df)
```

```python
#Convert Amount to double
df["Amount"] = df["Amount"].transform(lambda x: x * 2)

print(df)
```

```python
#Group + Transformation
df["Category_Average"] = df.groupby("Category")["Amount"].transform("mean")

print(df)
```

```python
#Group + transform with count
df["Category_Count"] = df.groupby("Category")["Amount"].transform("count")

print(df)
```
