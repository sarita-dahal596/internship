```python
import sqlite3

# Connect to database
connection = sqlite3.connect("college.db")

# Create cursor(used to send sql command to the database suchas create,insert,select
cursor = connection.cursor()

# Create Student table
cursor.execute("""
CREATE TABLE IF NOT EXISTS Student (
    student_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER
)
""")

# Create Enrollment table
cursor.execute("""
CREATE TABLE IF NOT EXISTS Enrollment (
    enrollment_id INTEGER PRIMARY KEY,
    student_id INTEGER,
    course TEXT,
    FOREIGN KEY (student_id) REFERENCES Student(student_id)
)
""")

# Insert students
cursor.execute(
    "INSERT OR IGNORE INTO Student VALUES (?, ?, ?)",
    (1, "Ram", 20)
)

cursor.execute(
    "INSERT OR IGNORE INTO Student VALUES (?, ?, ?)",
    (2, "Sita", 21)
)

# Insert enrollment
cursor.execute(
    "INSERT OR IGNORE INTO Enrollment VALUES (?, ?, ?)",
    (101, 1, "Python")
)

# Save changes
connection.commit()

# Display students
cursor.execute("SELECT * FROM Student")

for row in cursor.fetchall():
    print(row)

# Close connection
connection.close()
```

DATABASE MODEL

```python
#create engine connect python to the db
from sqlalchemy import create_engine, Column, Integer, String
from sqlalchemy.orm import declarative_base


engine = create_engine("sqlite:///college.db") #This line tells SQLAlchemy to connect to the SQLite database college.d

Base = declarative_base()

class StudentNew(Base):
    # New table name
    __tablename__ = "students"

    id = Column(Integer, primary_key=True)
    name = Column(String)
    age = Column(Integer)

# Create the table
Base.metadata.create_all(engine)

print("Table created")
```

```python
# Import session maker (used to create database sessions.)
#A session is what you use to interact with the database data—for example, to insert, read, update, and delete records.
from sqlalchemy.orm import sessionmaker

# Create session connected to the engine
Session = sessionmaker(bind=engine)

# Create session object
session = Session()

print("Session created")
```

```python
# Create student objects
student1 = StudentNew(name="alice", age=20)
student2 = StudentNew(name="babita", age=23)
student3 = StudentNew(name="daina", age=45)

# Add students
session.add_all([student1, student2, student3])

# Save data in database
session.commit()

print("Students added")
```

```python
# Get all students
students = session.query(StudentNew).all()

# Display students
for student in students:
    print(student.id, student.name, student.age)
```

one to many

```python
from sqlalchemy import Column, Integer, String, ForeignKey
from sqlalchemy.orm import relationship


# ONE side
class TeacherRelation(Base):
    __tablename__ = "teacher_relation"

    id = Column(Integer, primary_key=True)
    name = Column(String)
    subject = Column(String)
    age = Column(Integer)

    # One teacher can have many students
    students = relationship(
        "StudentRelation",
        back_populates="teacher"  #connect this relationship to the teacher relationship on the other model
    )


# MANY side
class StudentRelation(Base):
    __tablename__ = "student_relation"

    id = Column(Integer, primary_key=True)
    name = Column(String)
    age = Column(Integer)

    # Foreign key connects student to teacher
    teacher_id = Column(
        Integer,
        ForeignKey("teacher_relation.id")
    )

    # Each student belongs to one teacher
    teacher = relationship(
        "TeacherRelation",
        back_populates="students" #connect this relationship to the students relationship on the other model
    )


# Create tables
Base.metadata.create_all(engine)

print("Relationship tables created")
```

```python
# Create one teacher
teacher1 = TeacherRelation(
    name="Jack",
    subject="Python",
    age=40
)

# Create students
student1 = StudentRelation(
    name="Ram",
    age=20
)

student2 = StudentRelation(
    name="Sita",
    age=21
)

student3 = StudentRelation(
    name="Hari",
    age=22
)

# Give these students to teacher1
teacher1.students = [
    student1,
    student2,
    student3
]

# Add teacher
session.add(teacher1)

# Save teacher and related students
session.commit()

print("Teacher and students added")
#read teacher and student
# Find teacher
teacher = session.query(TeacherRelation).filter_by(
    name="Jack"
).first()

# Print teacher
print("Teacher:", teacher.name)

# Print teacher's students
for student in teacher.students:
    print(student.name, student.age)
```

```python
#accessing  relationship from student->teacher
student = session.query(StudentRelation).filter_by(
    name="Ram"
).first()

print("Student:", student.name)
print("Teacher:", student.teacher.name)
```

```python
# Imports
from sqlalchemy import create_engine, Column, Integer, String, ForeignKey
from sqlalchemy.orm import declarative_base, relationship, sessionmaker


# Connect to your existing database
engine = create_engine("sqlite:///college.db")


# Create Base
Base = declarative_base() # base parent of all your database model classes.


# EXISTING students table model
class StudentNew(Base):
    __tablename__ = "students"

    id = Column(Integer, primary_key=True)
    name = Column(String)
    age = Column(Integer)

    # One student has one card
    card = relationship(
        "StudentCard",
        back_populates="student", #connect this relationship to the student relationship on th other model
        uselist=False
    )


# NEW StudentCard model
class StudentCard(Base):
    __tablename__ = "student_cards"

    id = Column(Integer, primary_key=True)
    card_number = Column(String)

    # Connect to EXISTING students.id
    student_id = Column(
        Integer,
        ForeignKey("students.id"),
        unique=True
    )

    # Card belongs to one student
    student = relationship(
        "StudentNew",
        back_populates="card"
    )


# Creates student_cards if it doesn't exist.
# It will NOT recreate the existing students table.
Base.metadata.create_all(engine)


# Create session
Session = sessionmaker(bind=engine)
session = Session()

print("Models ready")
```

```python
#define base again
from sqlalchemy import (
    create_engine,
    Column,
    Integer,
    String,
    ForeignKey,
    Table
)

from sqlalchemy.orm import (
    declarative_base,
    relationship,
    sessionmaker
)

# Connect to database
engine = create_engine("sqlite:///college.db")

# Create Base
Base = declarative_base()

print("Base created")
```

```python
#define the middle table
student_course = Table(
    "student_course",
    Base.metadata,

    Column(
        "student_id",
        Integer,
        ForeignKey("students.id"),
        primary_key=True
    ),

    Column(
        "course_id",
        Integer,
        ForeignKey("courses.id"),
        primary_key=True
    )
)
```

```python
#Define Student
class StudentNew(Base):
    __tablename__ = "students"

    id = Column(Integer, primary_key=True)
    name = Column(String)
    age = Column(Integer)

    courses = relationship(
        "Course",
        secondary=student_course,
        back_populates="students"
    )
```

```python
#Define course
class Course(Base):
    __tablename__ = "courses"

    id = Column(Integer, primary_key=True)
    course_name = Column(String)

    students = relationship(
        "StudentNew",
        secondary=student_course,
        back_populates="courses"
    )
```

```python
Base.metadata.create_all(engine)

print("Tables ready")
```

```python
#create session
Session = sessionmaker(bind=engine)
session = Session()

print("Session ready")
```

```python
#assigning courses with subject
assignments = {
    1: ["Python", "Database"],
    2: ["Database", "AI"],
    3: ["Python"],
    4: ["Python", "AI"],
    5: ["Database"],
    6: ["AI","Python","Database"]
}
```

```python
#automatically assigns different courses to different students using the assignments dictionary we created earlier.
for student_id, course_names in assignments.items(): #Loop through the assignments

    student = session.get(StudentNew, student_id) #Find that student from database

    #Find the required courses
    courses = session.query(Course).filter(
        Course.course_name.in_(course_names)
    ).all()

    student.courses = courses #Assign courses to student

session.commit()
```
