enter exit

```python
class MyManager:

    def __enter__(self):  #__enter__() is automatically called when Python enters the with block.
        print("Context started")
        return self

    def __exit__(self, exc_type, exc_value, traceback):#When the with block finishes, Python automatically calls__exit__()
        print("Context ended")


with MyManager() as manager: #After __enter__() finishes, Python executes the code inside the with block.
    print("Inside context manager")
```

```python
class Student:
    def __enter__(self):
        print("student entered")
        return self

    def __exit__(self,exc_type,exc_value,traceback):
        print("student exited")

with Student() as student:
    print("hello  ")
```

```python
#custom file manager
class FileManager:

    def __enter__(self):
        self.file = open("data.txt", "w")
        return self.file

    def __exit__(self, exc_type, exc_value, traceback):
        self.file.close()
        print("File closed")


with FileManager() as file:
    file.write("Hello Python")
```

```python
#custom file manager
class Practice:
    def __enter__(self):
        self.file = open("students.txt","w")
        return self.file

    def __exit__(self,exc_type,exc_value,traceback):
        self.file.close()
        print("file stoped")

with Practice() as practice:
    practice.write("hello ram")

```

```python
#custom manager with constructor
class School:
    def __init__(self,name):
        self.name = name

    def __enter__(self):
        print("school name is:",self.name)

    def __exit__(self,exc_type,exc_value,traceback):
        print("BYE BYE :",self.name)

with School("sunrise school") as school:
    print("school started")
```

```python
#timer context manager
#This calculates how long a block of code takes.
import time
class Timer:
    def __enter__(self):
        self.start = time.time()

    def __exit__(self,exc_type,exc_value,traceback):
        self.end = time.time()
        print("time:",self.end-self.start)


with Timer():
    for i in range(1000000):
        pass
```

```python
class ErrorManager:

    def __enter__(self):
        print("Started")
        return self

    def __exit__(self, exc_type, exc_value, traceback):  #__exit__() receives the error
        #exc_type= ZeroDivisionError,exc_value = division by zero,traceback= information about where the error occurred
        if exc_type:  #Check whether an error occurred
            print("Error:", exc_value)

        print("Ended")
        return True  #The exception has been handled. Don't raise it again.


with ErrorManager():
    print(10 / 0)  #Instead of immediately stopping, Python passes information about this error to __exit__()
```

```python
class Calculator:
    def __enter__(self):
        print("calculator started")
        return self

    def add(self,a,b):
        return a+b

    def mul(self,a,b):
        return a*b

    def __exit__(self,exc_type,exc_value,traceback):
        print("calculator ended")

with Calculator() as calculator:
    print(calculator.add(1,2))
    print(calculator.mul(1,2))
```

```python
#transaction
class Transaction:
    def __enter__(self):
        print("Transaction started:")

    def __exit__(self,exc_type,exc_value,traceback):
        if exc_type:
            print("Error:", exc_value)
            print("Transaction ended")
            return True

        print("Transaction committed")


with Transaction():
    amount = 1000
    #print(10/0)
    print("The amount is:",amount)



```

contextlib

```python
from contextlib import contextmanager

# Create a custom context manager
@contextmanager
def my_manager():

    # Setup code - runs before entering the with block
    print("Started")

    # Give control to the with block
    yield

    print("Ended")

# Use the context manager
with my_manager():

    # This code runs at the yield position
    print("Working")
```

```python
from contextlib import contextmanager
@contextmanager
def student():
    print("Student entered")
    yield"alina"
    print("ended")

with student() as  name:
    print("Student",name)


```

```python
from contextlib import contextmanager
@contextmanager
def teacher():
    print("good morning")
    yield"teacher"

with teacher() as t:
    print("how are you",t)
```

```python


@contextmanager
def file_manager():

    # Open file
    file = open("data.txt", "w")

    try:
        # Give file to the with block
        yield file

    finally:
        # Close file even if error occurs
        file.close()
        print("File closed")


with file_manager() as file:
    # Write data into file
    file.write("learning python")
```

```python
from contextlib import contextmanager
@contextmanager
def manager():

    # open a file
    file = open("students.txt","w")
    try:
        #give file to the with block
        yield file

    finally:
        # close file even if errors occurs
        file.close()
        print("File closed")

with manager() as file:
     file.write("welcome students")


```

```python
#suppress() is a context manager from Python's contextlib module. It is used to ignore a specific exception and allow the program to continue.
from contextlib import suppress

with suppress(FileNotFoundError):
    file = open("abc.txt", "r")#If abc.txt does not exist Python suppresses (ignores) that specific error and moves to the next statement:

print("Program continues")
```

```python
from contextlib import suppress

# Ignore division by zero error
with suppress(ZeroDivisionError):
    result = 10 / 0
    print(result)

print("Program continues")
```

```python
from contextlib import suppress

# Ignore ValueError
with suppress(ValueError):
    number = int("hello")

print("Program continues")
```
