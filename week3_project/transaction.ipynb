Transaction Frequency Analysis


```python
#Transactions Per Customer
customer_frequency = (
    df.groupby("client_id")["id"]
    .count()
    .sort_values(ascending=False)
)

customer_frequency.head(10)
```

```python
#Top 10 Customers by Transaction Frequency.10-weight of the graph, 5-height of the graph
customer_frequency.head(10).plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Top 10 Customers by Transaction Frequency")
plt.xlabel("Client ID")
plt.ylabel("Number of Transactions")

plt.xticks(rotation=45)

plt.show()
```

Customer Spending Patterns

```python
#Total Spending Per Customer
customer_spending = (
    spending.groupby("client_id")["amount"]
    .sum()
    .sort_values(ascending=False)
)

customer_spending.head(10)
```

```python
#Top 10 Spending Customers. 10-weight of the graph, 5-height of the graph
customer_spending.head(10).plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Top 10 Customers by Spending")
plt.xlabel("Client ID")
plt.ylabel("Total Spending")

plt.xticks(rotation=45)

plt.show()
```
